Governança da gold — documentação, tags e lineage
Metadado, quando o consumidor é uma IA, deixa de ser documentação e vira **requisito funcional**. O `COMMENT` é literalmente o texto que o modelo lê para escolher qual coluna usar: coluna mal descrita é coluna usada errado, e o erro chega bonito, formatado e com número.
Três entregas aqui:
1. `COMMENT` em cada tabela e em **cada** coluna da `obt_voos` — descrevendo significado de negócio, não tipo de dado;
2. tags nas tabelas gold;
3. lineage do bronze até a OBT.

E um passo que **não** dá para pular: revisar as descrições geradas pela IA uma a uma, contra o dado. Uma delas está errada.

%md
## 1. Revisão: a descrição que a IA errou

Este foi o rascunho gerado para a métrica `minutos_recuperados`:

> *"Minutos que a etapa recuperou em voo. Valor positivo indica que o voo chegou
> adiantado."*

Soa perfeito. E está errado — e o jeito de saber não é reler a frase, é
MAGIC **perguntar para o dado**.

```python
%sql
CREATE OR REPLACE TABLE voos_brasil.gold.obt_voos AS
WITH base AS (
  SELECT
    v.icao_empresa,
    v.numero_voo,
    v.codigo_di,
    v.codigo_tipo_linha,
    v.icao_origem,
    v.icao_destino,
    v.partida_prevista,
    v.partida_prevista_data,
    v.partida_prevista_hora,
    v.partida_real,
    v.chegada_prevista,
    v.chegada_real,
    v.situacao_voo,
    v.codigo_justificativa,
    v.atraso_partida_min,
    v.atraso_chegada_min,
    v.minutos_recuperados,
    v._arquivo_origem,
    v._ingerido_em,
    e.sigla_iata        AS empresa_iata,
    COALESCE(e.razao_social, CONCAT('COMPANHIA NAO CADASTRADA ', v.icao_empresa)) AS nome_companhia,
    COALESCE(cdi.descricao, CONCAT('Codigo ', v.codigo_di, ' nao catalogado pela ANAC')) AS descricao_di,
    COALESCE(ctl.descricao, CONCAT('Codigo ', v.codigo_tipo_linha, ' nao catalogado pela ANAC')) AS descricao_tipo_linha,
    COALESCE(ao.nome, CONCAT('AEROPORTO FORA DO CADASTRO ANAC ', v.icao_origem)) AS nome_aeroporto_origem,
    ao.municipio        AS municipio_origem,
    ao.uf_nome          AS uf_origem,
    COALESCE(ad.nome, CONCAT('AEROPORTO FORA DO CADASTRO ANAC ', v.icao_destino)) AS nome_aeroporto_destino,
    ad.municipio        AS municipio_destino,
    ad.uf_nome          AS uf_destino
  FROM voos_brasil.silver.vra v
  LEFT JOIN voos_brasil.silver.empresas e
    ON v.icao_empresa = e.icao
  LEFT JOIN voos_brasil.silver.aerodromos ao
    ON v.icao_origem = ao.icao
  LEFT JOIN voos_brasil.silver.aerodromos ad
    ON v.icao_destino = ad.icao
  LEFT JOIN voos_brasil.silver.codigos_operacao cdi
    ON cdi.dominio = 'codigo_di' AND v.codigo_di = cdi.codigo
  LEFT JOIN voos_brasil.silver.codigos_operacao ctl
    ON ctl.dominio = 'codigo_tipo_linha' AND v.codigo_tipo_linha = ctl.codigo
),
metricas AS (
  SELECT
    *,
    CASE WHEN atraso_partida_min IS NOT NULL
           AND (atraso_partida_min < -120 OR atraso_partida_min > 1440)
         THEN true ELSE false END AS atraso_fora_de_faixa
  FROM base
)
SELECT
  icao_empresa,
  empresa_iata,
  nome_companhia,
  numero_voo,
  codigo_di,
  descricao_di,
  codigo_tipo_linha,
  descricao_tipo_linha,
  CASE WHEN codigo_tipo_linha IN ('N','C') THEN 'Domestico'
       WHEN codigo_tipo_linha IN ('I','G') THEN 'Internacional'
  END                                                       AS escopo_voo,
  icao_origem,
  nome_aeroporto_origem,
  municipio_origem,
  uf_origem,
  CASE WHEN icao_origem LIKE 'SB%' THEN 'Brasil' ELSE 'Exterior' END AS pais_origem,
  icao_destino,
  nome_aeroporto_destino,
  municipio_destino,
  uf_destino,
  CASE WHEN icao_destino LIKE 'SB%' THEN 'Brasil' ELSE 'Exterior' END AS pais_destino,
  CONCAT(icao_origem, ' - ', icao_destino)                   AS rota_icao,
  CONCAT(COALESCE(municipio_origem, icao_origem), ' - ',
         COALESCE(municipio_destino, icao_destino))           AS rota_municipios,
  partida_prevista,
  partida_prevista_data,
  partida_prevista_hora,
  HOUR(partida_prevista)                                     AS hora_partida_prevista,
  CASE dayofweek(partida_prevista)
    WHEN 1 THEN 'domingo'
    WHEN 2 THEN 'segunda-feira'
    WHEN 3 THEN 'terca-feira'
    WHEN 4 THEN 'quarta-feira'
    WHEN 5 THEN 'quinta-feira'
    WHEN 6 THEN 'sexta-feira'
    WHEN 7 THEN 'sabado'
  END                                                       AS dia_semana,
  CAST(date_trunc('month', partida_prevista) AS DATE)       AS mes_referencia,
  partida_real,
  chegada_prevista,
  chegada_real,
  CASE WHEN atraso_fora_de_faixa THEN NULL ELSE atraso_partida_min  END AS atraso_partida_min,
  CASE WHEN atraso_fora_de_faixa THEN NULL ELSE atraso_chegada_min  END AS atraso_chegada_min,
  CASE WHEN atraso_fora_de_faixa THEN NULL ELSE minutos_recuperados END AS minutos_recuperados,
  atraso_fora_de_faixa,
  CASE WHEN atraso_partida_min IS NULL OR atraso_fora_de_faixa THEN NULL
       WHEN atraso_partida_min <= 15 THEN true
       ELSE false END                                       AS partida_pontual,
  CASE WHEN atraso_chegada_min IS NULL OR atraso_fora_de_faixa THEN NULL
       WHEN atraso_chegada_min <= 15 THEN true
       ELSE false END                                       AS chegada_pontual,
  situacao_voo,
  CASE WHEN situacao_voo = 'REALIZADO' THEN true ELSE false END AS voo_realizado,
  CASE WHEN situacao_voo = 'CANCELADO' THEN true ELSE false END AS voo_cancelado,
  codigo_justificativa,
  _arquivo_origem,
  _ingerido_em,
  current_timestamp()                                       AS _processado_em
FROM metricas
```

```python


display(spark.sql("""
    SELECT
      COUNT(*)                                                      AS recuperou_algum_minuto,
      SUM(CASE WHEN atraso_chegada_min <= 0 THEN 1 ELSE 0 END)      AS chegou_adiantado_ou_no_horario,
      SUM(CASE WHEN atraso_chegada_min >  0 THEN 1 ELSE 0 END)      AS chegou_atrasado_mesmo_assim,
      SUM(CASE WHEN atraso_chegada_min > 15 THEN 1 ELSE 0 END)      AS chegou_atrasado_mais_de_15
    FROM voos_brasil.gold.obt_voos
    WHERE minutos_recuperados > 0
"""))
```

```python
%sql
DESCRIBE voos_brasil.gold.obt_voos;
```

**164.895 voos recuperaram tempo no ar e mesmo assim chegaram atrasados** — 75.082 deles com mais de 15 minutos de atraso. A descrição da IA teria feito qualquer pessoa (e qualquer LLM) concluir o contrário.

O certo é: *positivo significa que a etapa chegou **menos atrasada do que saiu**.
Não quer dizer que chegou no horário.*
A diferença entre as duas frases é a diferença entre "recuperou" e "resolveu". A segunda métrica, `chegada_pontual`, é quem responde se resolveu.

Outras duas afirmações do rascunho que a revisão também derrubou:
| rascunho | o dado diz |
|---|---|
| "`partida_pontual = false` inclui os voos cancelados" | cancelado tem `partida_pontual` NULL: 29.140 de 29.140 |
| "`mes_referencia` é o mês de todo voo" | é NULL em 30.798 voos, os que não têm horário previsto |

```python
display(spark.sql("""
    SELECT situacao_voo,
           COUNT(*)                                                AS voos,
           SUM(CASE WHEN atraso_partida_min IS NULL THEN 1 ELSE 0 END) AS partida_pontual_null
    FROM voos_brasil.gold.obt_voos GROUP BY situacao_voo
"""))
```

## 3. Comentários do fato e da dimensão
A OBT é a tabela que a IA lê, mas o fato e a dimensão continuam sendo lidos por gente. Eles herdam as mesmas descrições onde a coluna é a mesma.


```python
%sql
CREATE OR REPLACE TABLE voos_brasil.gold.dim_aeroporto AS
WITH aeroportos AS (
  SELECT icao_origem AS icao_aeroporto FROM voos_brasil.silver.vra
  UNION
  SELECT icao_destino FROM voos_brasil.silver.vra
)
SELECT
  a.icao_aeroporto,
  COALESCE(ae.nome, CONCAT('AEROPORTO FORA DO CADASTRO ANAC ', a.icao_aeroporto)) AS nome_aeroporto,
  ae.municipio AS municipio_aeroporto,
  ae.uf_nome   AS uf_aeroporto,
  CASE WHEN a.icao_aeroporto LIKE 'SB%' THEN 'Brasil' ELSE 'Exterior' END AS pais_aeroporto,
  ae.icao IS NOT NULL AS no_cadastro_anac,
  current_timestamp() AS _processado_em
FROM aeroportos a
LEFT JOIN voos_brasil.silver.aerodromos ae
  ON a.icao_aeroporto = ae.icao;

CREATE OR REPLACE TABLE voos_brasil.gold.fato_voos AS
SELECT
  o.icao_empresa,
  o.empresa_iata,
  o.nome_companhia,
  o.numero_voo,
  o.codigo_di,
  o.descricao_di,
  o.codigo_tipo_linha,
  o.descricao_tipo_linha,
  o.escopo_voo,
  o.icao_origem,
  o.icao_destino,
  o.rota_icao AS rota,
  e.origem_cadastro AS cadastro_companhia,
  o.partida_prevista,
  o.partida_prevista_data,
  o.partida_prevista_hora,
  o.hora_partida_prevista,
  o.dia_semana,
  o.mes_referencia,
  o.partida_real,
  o.chegada_prevista,
  o.chegada_real,
  o.atraso_partida_min,
  o.atraso_chegada_min,
  o.minutos_recuperados,
  o.atraso_fora_de_faixa,
  o.partida_pontual,
  o.chegada_pontual,
  o.situacao_voo,
  o.voo_realizado,
  o.voo_cancelado,
  o.codigo_justificativa,
  o._arquivo_origem,
  o._ingerido_em,
  current_timestamp() AS _processado_em
FROM voos_brasil.gold.obt_voos o
LEFT JOIN voos_brasil.silver.empresas e
  ON o.icao_empresa = e.icao;
```

```python
COMENTARIOS_DIM = {
    "icao_aeroporto":      "Codigo ICAO do aeroporto. Chave da dimensao, serve tanto para origem quanto para destino do fato.",
    "nome_aeroporto":      "Nome do aeroporto. Traz fallback textual com o codigo quando o aeroporto nao esta no cadastro da ANAC.",
    "municipio_aeroporto": "Municipio onde o aeroporto esta localizado. Vazio para aeroporto estrangeiro.",
    "uf_aeroporto":        "Unidade federativa por extenso, como a ANAC publica. Nao e a sigla.",
    "pais_aeroporto":      "Brasil ou Exterior, deduzido do prefixo ICAO. Regra de negocio criada na gold.",
    "no_cadastro_anac":    "Verdadeiro quando o aeroporto existe no cadastro de aerodromos publicos da ANAC. Falso e o esperado para aeroporto estrangeiro, e nao indica erro.",
    "_processado_em":      "Auditoria: momento da construcao da dimensao.",
}

COMENTARIOS_FATO = dict(COMENTARIOS_OBT)
COMENTARIOS_FATO["icao_origem"]  = "Codigo ICAO do aeroporto de partida. Chave para gold.dim_aeroporto."
COMENTARIOS_FATO["icao_destino"] = "Codigo ICAO do aeroporto de chegada. Chave para gold.dim_aeroporto."
COMENTARIOS_FATO["rota"] = "Rota no formato ORIGEM - DESTINO usando codigos ICAO."
COMENTARIOS_FATO["cadastro_companhia"] = "De qual cadastro da ANAC veio a companhia: nacional ou estrangeira. Nulo quando o codigo nao tem cadastro."
COLUNAS_FATO = [c for c in COMENTARIOS_FATO if c not in (
    "nome_aeroporto_origem", "municipio_origem", "uf_origem", "pais_origem",
    "nome_aeroporto_destino", "municipio_destino", "uf_destino", "pais_destino",
    "rota_icao", "rota_municipios")]

for coluna in COLUNAS_FATO:
    spark.sql(f"ALTER TABLE voos_brasil.gold.fato_voos ALTER COLUMN {coluna} COMMENT '{COMENTARIOS_FATO[coluna]}'")
print(f"{len(COLUNAS_FATO)} colunas comentadas em gold.fato_voos")

for coluna, comentario in COMENTARIOS_DIM.items():
    spark.sql(f"ALTER TABLE voos_brasil.gold.dim_aeroporto ALTER COLUMN {coluna} COMMENT '{comentario}'")
print(f"{len(COMENTARIOS_DIM)} colunas comentadas em gold.dim_aeroporto")
```

## 6. Lineage — do arquivo cru até a OBT
Ninguém escreveu uma linha de código de rastreabilidade. O Unity Catalog registrou sozinho quem leu o quê para escrever o quê.



```python
display(spark.sql("""
    SELECT 
        COALESCE(NULLIF(source_table_full_name, ''), '(arquivo no volume)') AS origem,
        target_table_full_name AS destino
    FROM system.access.table_lineage
    WHERE target_table_full_name IS NOT NULL
      AND target_table_full_name LIKE 'voos_brasil.%'
      AND event_date >= current_date() - 7
    GROUP BY 1, 2
    ORDER BY destino, origem
"""))
```

É esse grafo que responde, em dez segundos, a pergunta mais cara de um time de
dados: *"se eu mexer aqui, o que quebra lá na frente?"*.
E ele responde também a pergunta do compliance, na direção contrária: *"esse número no relatório veio de onde?"* — `obt_voos` ← `fato_voos` ← `silver.vra` ← `bronze.vra` ← arquivo CSV da ANAC no volume, com data e hora de cada passo.
